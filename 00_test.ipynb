```python
# if elif else statement for marks of the students
# 95 and above - grade A, 80 to 94 - grade B, 75 to 79 - grade C, 70 to 74 - grade D, below 70 - fail
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("You got grade A")
elif marks >= 80:
    print("You got grade B")
elif marks >= 75:
    print("You got grade C")
elif marks >= 70:
    print("You got grade D")
else:
    print("You failed")
```
