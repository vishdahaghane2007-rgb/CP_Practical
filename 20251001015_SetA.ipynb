```python
# Initial concentration (mol/L)
# Rate constant (1/min)
# Time step (mi
C0 = 2.0
k = 0.08       
dt = 1.0       

C = C0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")

for t in range(11):
    conversion = (C0 - C) / C0 * 100
    print(f"{t}\t\t{C:.3f}\t\t\t{conversion:.2f}")

    # Euler method
    C = C - k * C * dt
```

using math.exp(), calculate the exact value C0e^(-kt) at v10 min and priont the absoulte error of the euler method . give a code for this.

```python
# the absolute error between the Euler method

C0 = 2.0
k = 0.08
dt = 1.0
C_euler = C0

for t in range(10):
    C_euler = C_euler - k * C_euler * dt

C_exact = C0 * math.exp(-k * 10)
error = abs(C_exact - C_euler)

print("Euler concentration at 10 min =", round(C_euler, 4), "mol/L")
print("Exact concentration at 10 min =", round(C_exact, 4), "mol/L")
print("Absolute error =", round(error, 4), "mol/L")
```

using a while loop, repeat the simulation until conversion reaches 60% or time reaches 100 min. print the time needed and the final concentrations. give a code.

```python
# Initial concentration (mol/L)
# Rate constant (1/min)
# Time step (min)
C0 = 2.0       
k = 0.08       
dt = 1.0      

C = C0
time = 0

while time < 100 and (C0 - C) / C0 * 100 < 60:
    C = C - k * C * dt
    time = time + dt

conversion = (C0 - C) / C0 * 100

print("Time needed:", time, "min")
print("Final concentration:", round(C, 3), "mol/L")
print("Final conversion:", round(conversion, 2), "%")
```

```python
label = input("Enter batch label: ")

if label == "":
    print("The label is empty.")
else:
    digit_count = 0
    digit_sum = 0

    for character in label:
        if character.isdigit():
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The label contains no digits.")
    else:
        print("Number of digits:", digit_count)
        print("Sum of digits:", digit_sum)
```

QUESTION NO. 03 ADDITIVE SCREENING AND SAMPLING SCHEDULE

PART A:  a laboratory has n candiadte additives and wants to test r of them togther in one formulation (order does not matter). write a recursive function factorial(n)(remeber the base case for 0!) and a function combination(n,r)=n!/r!(n-r)! that uses it. reject invaild input (negative values, r>n). print the number of possible formulations for n=6, n=3, and show the message for n=5,r=7. Give a code



```python
#the number of possible combinations of n items taken r at a time (nCr) using the formula nCr = n! / (r! * (n - r)!)
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0:
        return "Error: values cannot be negative."
    if r > n:
        return "Error: r cannot be greater than n."

    return factorial(n) // (factorial(r) * factorial(n - r))


print("Combinations for n = 6, r = 3:", combination(6, 3))
print("Combinations for n = 3, r = 3:", combination(3, 3))
print("For n = 5, r = 7:", combination(5, 7))
```

part B: during a reaction, sample are taken at times = fibonacci number*2 min. write an iterative function fib_iter(n) that returns a list of the frist n fibonacci numbers, and a recursive function fib_rec(k) that returns the kth fibonacci number (F0= 0, F1= 1). for n=10, check that both function agree and print a table : sample no. |fibonacci tag| sampling time(min). Give a code


```python
# part B 
def fib_iter(n):
    numbers = []

    for i in range(n):
        numbers.append(fib_rec(i))

    return numbers


def fib_rec(k):
    if k == 0:
        return 0
    elif k == 1:
        return 1
    else:
        return fib_rec(k - 1) + fib_rec(k - 2)


n = 10
fibonacci_numbers = fib_iter(n)

print("Both functions agree:", all(
    fibonacci_numbers[i] == fib_rec(i) for i in range(n)
))

print("\nSample No.\tFibonacci Tag\tSampling Time (min)")

for i in range(n):
    time = fibonacci_numbers[i] * 2
    print(f"{i + 1}\t\t{fibonacci_numbers[i]}\t\t{time}")
```

print the mean and std dev of run1 and run 2 and state which run shows more variation

```python
import pandas as pd

run1 = pd.Series([1.00, 0.74, 0.55, 0.41, 0.30, 0.22])
run2 = pd.Series([1.00, 0.61, 0.37, 0.22, 0.14, 0.08])

mean1 = run1.mean()
std1 = run1.std()

mean2 = run2.mean()
std2 = run2.std()

print("Run 1 (Low temperature)")
print("Mean =", round(mean1, 3))
print("Standard deviation =", round(std1, 3))

print("\nRun 2 (High temperature)")
print("Mean =", round(mean2, 3))
print("Standard deviation =", round(std2, 3))

if std1 > std2:
    print("\nRun 1 shows more variation.")
else:
    print("\nRun 2 shows more variation.")
```

 plot  both runs on one graph (time on x-axis ) with title, axis labels with units, legend and grid.

nitrogen gas is compressed slowly in piston-cylinder at constant temperature: volume_l as 1, 2,3,4,5 and pressure_bar as 4.00 ,2.03,1.31,1.02,0.81. plot pressure vs.volume as ascatter plot. calculate k as the mean of P*V and draw the reference curve P+k/V on the same graph. add title, labels with units and legend. 

```python
import matplotlib.pyplot as plt

volume_l = [1, 2, 3, 4, 5]
pressure_bar = [4.00, 2.03, 1.31, 1.02, 0.81]

# Calculate k = mean(P * V)
pv_values = []

for v, p in zip(volume_l, pressure_bar):
    pv_values.append(p * v)

k = sum(pv_values) / len(pv_values)

# Reference pressure: P = k / V
reference_pressure = []

for v in volume_l:
    reference_pressure.append(k / v)

print("Mean value of k =", round(k, 3), "bar L")

plt.scatter(volume_l, pressure_bar, label="Experimental data")
plt.plot(volume_l, reference_pressure, label="Reference curve: P = k/V")

plt.title("Nitrogen Gas Compression")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)

plt.show()
```
